```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
metadata_path = "/path/to/AstroBF_fastq/RN22346/RN22346.xlsx"
metadata = pd.read_excel(metadata_path)
```

```python
metadata['category'] = metadata["Sample Genotype"] + "_" + metadata["Sample Treatment"]
```

```python
df_genes = pd.read_csv("/path/to/AstroBF_fastq/RN22346/230106_A01366_0335_BHJJHNDSX5/genes_preprocessing/df_genes_filtered_normalized.csv", index_col=0)
```

```python
df_genes.columns = metadata.reset_index().set_index("limsid").loc[df_genes.columns]['category'].values
```

```python
df_genes
```

```python
treatments_cell_line = {'ctrl_basal': [],
                        'ctrl_A1 low': [],
                        'ctrl_A1 high': [],
                        'vcp_basal': [],
                        'vcp_A1 low': [],
                        'vcp_A1 high': [],}


for new_cat in treatments_cell_line.keys():
    for treatment in df_genes.columns:
        s = new_cat.split("_")[0]
        t = new_cat.split("_")[1]
        if s == 'ctrl':
            if 'ctrl' in treatment and t in treatment:
                treatments_cell_line[new_cat].append(treatment)
        else:
            if 'ctrl' not in treatment and t in treatment:
                treatments_cell_line[new_cat].append(treatment)
```

```python
treatments_cell_line
```

```python
df_genes_grouped = df_genes.copy()
for cat, treatments in treatments_cell_line.items():
    df_genes_grouped[cat] = df_genes[treatments].mean(axis=1)
    
df_genes_grouped = df_genes_grouped[list(treatments_cell_line.keys())]
```

```python
df_genes_grouped
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
l = linkage(df_genes_grouped.T, method='complete', metric='euclidean')
gif, ax = plt.subplots(1, 1, figsize=(10, 5))
d=dendrogram(l, ax=ax)
plt.xticks(rotation=90);
plt.title(f"Hierarchical clustering \n method=complete; metric=euclidean", weight='bold', fontsize=15);

# Set original column names as xticks
ax.set_xticklabels(df_genes_grouped.columns[d['leaves']], rotation=90)
```

```python
from scipy.stats import spearmanr
from scipy.spatial.distance import squareform
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
# Example: df_genes_grouped is (genes x samples), so we transpose
data = df_genes_grouped.T  

# Step 1: Spearman correlation between samples
corr, _ = spearmanr(data, axis=1)  # correlation matrix (samples x samples)

# Step 2: Convert correlation to distance
dist = 1 - corr  

# Step 3: Convert to condensed distance format for linkage
dist_condensed = squareform(dist, checks=False)

# Step 4: Perform clustering
l = linkage(dist_condensed, method='complete')

# Step 5: Plot dendrogram
fig, ax = plt.subplots(1, 1, figsize=(10, 5))
d = dendrogram(l, ax=ax)
plt.xticks(rotation=90)
plt.title("Hierarchical clustering\nmethod=complete; metric=spearman", weight='bold', fontsize=15)

# Set original sample names
ax.set_xticklabels(df_genes_grouped.columns[d['leaves']], rotation=90)
plt.show()
```

```python
l = linkage(df_genes.T, method='average', metric='euclidean')
gif, ax = plt.subplots(1, 1, figsize=(10, 5))
d=dendrogram(l, ax=ax)
plt.xticks(rotation=90);
plt.title(f"Hierarchical clustering \n method=average; metric=euclidean", weight='bold', fontsize=15);

# Set original column names as xticks
ax.set_xticklabels(df_genes.columns[d['leaves']], rotation=90)
```

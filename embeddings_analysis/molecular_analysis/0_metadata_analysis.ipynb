```python
import pandas as pd
```

```python
metadata_path = "path/to/AstroBF_fastq/RN22346/RN22346.xlsx"
metadata = pd.read_excel(metadata_path)
```

```python
metadata['category'] = metadata["Sample Genotype"] + "_" + metadata["Sample Treatment"]
metadata = metadata[metadata['Sample Treatment'].isin(["basal", "A1 low", "A1 high"])]
metadata = metadata[~metadata['Sample Genotype'].str.contains("ncrm")]
```

```python
metadata
```

```python
for id in metadata['limsid']:
    print(id)
```

```python
for genotype in metadata['Sample Genotype'].unique():
    print(metadata[metadata['Sample Genotype'] == genotype]['Sample Treatment'].value_counts())
```

For each genotype, we have a basal, A1 low and A1 high.

```python
# Create the treatment dictionnary
treatments = {}

for treatment in metadata['category'].unique():
    treatments[treatment] = metadata[metadata['category'] == treatment]['limsid'].tolist()
```

```python
for treatment, samples in treatments.items():
    print(f"{treatment}: {samples[0]}")
```
